```sql
-- play count aggregate by artist
SELECT
    artist_name,
    COUNT(*) AS artist_play_count
FROM spotify_project.gold.streaming_history
GROUP BY artist_name
ORDER BY artist_play_count DESC;


-- play count aggregate by artist and track
SELECT
    artist_name,
    track_name,
    COUNT(track_name) AS track_play_count
FROM spotify_project.gold.streaming_history
GROUP BY artist_name, track_name
ORDER BY track_play_count DESC;

-- play count aggregate by year
SELECT
    EXTRACT(YEAR FROM date_played) AS year,
    COUNT(*) AS play_count
FROM spotify_project.gold.streaming_history
GROUP BY year
ORDER BY year;

-- many genres are essentially duplicated (e.g. k-pop, kpop girl group), TODO: figure out how to handle this
CREATE OR REPLACE VIEW genres AS (
    SELECT 
        artist.artist_name,
        explode(genres) as genre 
    FROM spotify_project.gold.artist artist JOIN spotify_project.gold.streaming_history streaming_history ON artist.artist_name = streaming_history.artist_name  
);

-- aggregate by year and genre
-- SELECT
--     EXTRACT(YEAR FROM date_played) AS year,
--     genre,
--     COUNT(*) AS genre_play_count
-- FROM genres JOIN spotify_project.gold.streaming_history USING(artist_name)
-- GROUP BY year, genre
-- ORDER BY year, genre_play_count DESC;


-- aggregate by genre
SELECT 
    genre,
    COUNT(genre) AS genre_play_count
FROM genres
GROUP BY genre
ORDER BY genre_play_count DESC;

-- aggregate by platform
SELECT platform, COUNT(platform) AS platform_play_count FROM (
    SELECT
    CASE WHEN LOWER(platform) LIKE '%windows%' OR platform LIKE '%OS X%'THEN 'Desktop'
         WHEN LOWER(platform) LIKE '%ios%' THEN 'Mobile'
         ELSE 'Unknown' END AS platform
    FROM spotify_project.gold.streaming_history
    ) platform_plays
GROUP BY platform

```
