```sql
DROP TEMPORARY VARIABLE IF EXISTS sql_query;

DECLARE sql_query STRING;

SET VAR sql_query = (
  SELECT CONCAT(
    'CREATE OR REPLACE TABLE spotify_project.gold.streaming_history AS ',
    CONCAT_WS(
      ' UNION ALL ',
      COLLECT_LIST('SELECT 
        artist_name,
        track_name,
        album_name,
        play_time,
        date_played,
        platform,
        reason_start,
        reason_end,
        shuffle,
        skipped
      FROM ' || table_catalog || '.' || table_schema || '.' || table_name)
    )
  )
  FROM information_schema.tables
  WHERE table_schema = 'silver' AND table_name LIKE 'streaming_history_%'
);

EXECUTE IMMEDIATE sql_query;
```
