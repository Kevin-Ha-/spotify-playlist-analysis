```python
SOURCES_PATH = "/Volumes/spotify_project/bronze/user_play_history/"
files = dbutils.fs.ls(SOURCES_PATH)

json_files = [f for f in files if f.name.endswith(".json")]

for file in json_files:
    filename = file.name.replace('.json', '')
    
    print(f"Processing file {filename} -> spotify_project.bronze.{filename}")
    df = spark.read.option("multiline", "true").json(file.path)
    df.write.mode("overwrite").saveAsTable(f"spotify_project.bronze.{filename}")
```
