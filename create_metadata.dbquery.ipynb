```sql
CREATE SCHEMA IF NOT EXISTS spotify_project.metadata;

CREATE TABLE IF NOT EXISTS spotify_project.metadata.pipeline_runs (
  run_id STRING,
  start_time TIMESTAMP,
  end_time TIMESTAMP,
  status STRING,
  records_read INT,
  records_written INT,
  records_rejected INT
);

CREATE TABLE IF NOT EXISTS spotify_project.metadata.pipeline_watermarks (
  pipeline_name STRING,
  source STRING,
  last_success_run_id STRING,
  watermark_column STRING,
  watermark_value DATETIME,
  updated_at DATETIME
)
```
